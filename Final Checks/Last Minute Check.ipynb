```python
ten_docs_ret = "/home/ubuntu/RAG Pipeline/March 27 Chunks/1024-256-10-retrieved_chunks.json"
ten_cits_ret = "/home/ubuntu/RAG Pipeline/March 27 Citations/1024-256-10-retrieved_citations.json"

twenty_docs_ret = "/home/ubuntu/RAG Pipeline/March 27 Chunks/1024-256-20-retrieved_chunks.json"
twenty_cits_ret = "/home/ubuntu/RAG Pipeline/March 27 Citations/1024-256-20-retrieved_citations.json"

full_path = "/home/ubuntu/[March 27, 2025] Final Dataset/sampled_final_cleaned_acl_global_context_dataset_eval.csv"
```

```python
import json
import pandas as pd

def load_json(file_path):
    """Load JSON data from file."""
    with open(file_path, 'r', encoding='utf-8') as f:
        print(f"JSON file loaded from {file_path}")
        return json.load(f)

ten_cits = load_json(ten_cits_ret)
twenty_cits = load_json(twenty_cits_ret)

full_df = pd.read_csv(full_path)

citation_targets = full_df["masked_token_target"].tolist()

top_k = 10
```

```python
import numpy as np

recall_scores = []
mrr_scores = []
ndcg_scores = []

for i in range(len(citation_targets)):

    retrieved_citation_items = ten_cits[i]
    citation_target = citation_targets[i]

    # Keep only the highest-ranked document for each unique citation_item
    unique_citation_items = [] # List to track unique citations

    for citation in retrieved_citation_items:

        if citation and citation not in unique_citation_items:
            unique_citation_items.append(citation) # Track this citation
            
        if len(unique_citation_items) == top_k:  # Stop once we have enough unique results
            break

    citation_target = citation_target.strip()
    citation_set = {citation_target}  # Use set for O(1) lookup

    found = False

    for index, citation in enumerate(unique_citation_items):

        if citation.strip() in citation_set:
            recall_scores.append(1)
            mrr_scores.append(1 / (index + 1))
            ndcg_scores.append(1 / np.log2(index + 2))
            found = True
            break

    if not found:
        recall_scores.append(0)
        mrr_scores.append(0)
        ndcg_scores.append(0)

print("Overall metrics for 10 Documents")
print(f"Recall: {np.mean(recall_scores):.4f}")
print(f"MRR: {np.mean(mrr_scores):.4f}")
print(f"NDCG: {np.mean(ndcg_scores):.4f}")
```
