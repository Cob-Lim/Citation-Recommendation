```python
import chromadb

# Initialize ChromaDB client
client = chromadb.PersistentClient(path="/home/ubuntu/Infly-Chroma-Bread Pipeline/ChromaDB Storage")

# Get collection names
collection_names = client.list_collections()

# Print collection names
for name in collection_names:
    print(f"Collection Name: {name}")
```

```python
for name in collection_names:
    try:
        coll = client.get_collection(name)
        count = len(coll.get()["ids"])  # Get total documents
        print(f"Collection '{name}' has {count} documents")
    except Exception as e:
        print(f"Error accessing collection '{name}': {e}")
```

```python
to_delete = "150-75_vectorstore"
client.delete_collection(to_delete)
print(f"Deleted collection: {to_delete}")
```

### Clean json

```python
import json

# Step 1: Load raw JSON (with utf-8-sig to handle BOM)
with open("/home/ubuntu/[March 24, 2025] Final Dataset/full_text_sample_cleaned_global_eval_context_for_rag.json") as f:
    raw_data = json.load(f)

# Step 2: Check keys of first few entries
for i, entry in enumerate(raw_data[:3]):
    print(f"Entry {i} keys: {list(entry.keys())}")

# Step 3: Optional — filter out bad entries (e.g., missing full_text)
cleaned_data = [entry for entry in raw_data if 'full_text' in entry]

print(f"Cleaned entries: {len(cleaned_data)} / {len(raw_data)}")

# Step 4: Save to new file (with standard UTF-8, no BOM)
with open("cleaned_dataset.json", "w", encoding="utf-8") as f:
    json.dump(cleaned_data, f, ensure_ascii=False, indent=2)
```
